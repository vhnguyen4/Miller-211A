```python
import json
import textwrap

with open("speeches/Abraham_Lincoln_1854-10-16T13_03_58-04_56_October_16,_1854__At_Peoria,_Illinois.json", "r", encoding="utf-8") as f:
    data = json.load(f)

print(data["title"])
print()

paragraphs = data["transcript"].split("\n")

for paragraph in paragraphs:
    print(textwrap.fill(paragraph.strip(), width=140))
    print()
```
