```python
#set up
import pandas as pd
from pathlib import Path
import re
```

```python
folder_path = Path("./python_speeches")
president_names = []
for file_path in folder_path.glob("*.txt"):
    string = file_path.name
    president_name = re.split(r'\d', string, maxsplit=1)[0].rstrip(" _-")
    president_names.append(president_name)
    
president_df = pd.DataFrame(president_names, columns=["president_name"])
print(president_df)
```

```python
#Using president_df to hold data
president_statistics = president_df.groupby('president_name').agg(speeches = ('president_name', 'count'))
```

```python
#Not entirely sure why this works. Ask when talking to Tom
def keyword_search(word):
    folder_path = Path("./python_speeches")
    total_count = 0
    pattern = re.compile(rf"\b{re.escape(word)}\b", re.IGNORECASE)
    
    for file_path in folder_path.glob("*.txt"):
        text = file_path.read_text(encoding="utf-8")
        total_count += len(pattern.findall(text))
        
    return total_count
```
